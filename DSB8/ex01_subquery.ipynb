```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect('../data/checking-logs.sqlite')

# Получаем схему таблицы checker
schema_query = "PRAGMA table_info(checker);"
schema = pd.io.sql.read_sql(schema_query, conn)
print("Схема таблицы checker:")
print(schema)
```

```python
# Получаем первые 10 строк таблицы checker для ознакомления
first_10_query = "SELECT * FROM checker LIMIT 10;"
first_10_rows = pd.io.sql.read_sql(first_10_query, conn)
print("\nПервые 10 строк таблицы checker:")
print(first_10_rows)
```

## Основной запрос с подзапросом для подсчета отфильтрованных строк

```python
checkers = """
SELECT COUNT(*) as cnt
FROM pageviews
WHERE EXISTS (
    SELECT 1
    FROM checker
    WHERE checker.uid = pageviews.uid
        AND checker.status == 'ready'
        AND checker.numTrials = 1
        AND checker.labname IN ('laba04', 'laba04s', 'laba05', 'laba06', 'laba06s', 'project1')
);
"""

result = pd.io.sql.read_sql(checkers, conn)
```
