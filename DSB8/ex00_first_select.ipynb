```python
import pandas as pd
import sqlite3
```

```python
conn = sqlite3.connect('../data/checking-logs.sqlite')
```

```python
# Получаем схему таблицы pageviews
schema_query = "PRAGMA table_info(pageviews);"
schema = pd.io.sql.read_sql(schema_query, conn)
print("Схема таблицы pageviews:")
print(schema)
```

```python
# Получаем первые 10 строк для ознакомления
first_10_query = "SELECT * FROM pageviews LIMIT 10;"
first_10_rows = pd.io.sql.read_sql(first_10_query, conn)
print("\nПервые 10 строк таблицы pageviews:")
print(first_10_rows)
```

## Основной запрос: получение отфильтрованных данных

```python
query = """
SELECT uid, datetime 
FROM pageviews 
WHERE uid LIKE 'user_%' 
ORDER BY uid ASC;
"""

pageviews = pd.io.sql.read_sql(query, conn, index_col='datetime', parse_dates=['datetime'])
```
