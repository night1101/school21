RATINGS - ТАЙНЫ ПОЛЬЗОВАТЕЛЬСКИХ ОЦЕНОК

Таймлайн киноактивности — Когда мир смотрит кино?
Давайте отправимся в путешествие во времени! Узнаем, в какие годы пользователи были особенно активны и оставляли больше всего оценок.
Метод: dist_by_year() - возвращает словарь {год: количество оценок}, отсортированный по году.

```python
%timeit
print("РАСПРЕДЕЛЕНИЕ ОЦЕНОК ПО ГОДАМ")

yearly_dist = ratings_movies.dist_by_year()
if yearly_dist:
    for year, count in list(yearly_dist.items())[-15:]:
        bar = '█' * (count // 10)
        print(f"{year}: {count:4d} оценок {bar}")
else:
    print("Нет данных")
```

Анализ распределения по годам:

-Снижение активности - заметно уменьшение количества оценок в последние годы
-Пиковые годы- наиболее активные периоды оценивания 
-Данные извлекаются из timestamp (секунды с 1970 года)

ЛУЧШИЕ В ЖАНРЕ — Какой фильм в вашем любимом жанре получил самые высокие оценки?

Вы когда-нибудь задумывались, какой фильм в жанре драмы получил самые высокие оценки? А в комедии? Теперь у нас есть ответ! Новая функция `top_by_genre()` поможет найти настоящие жемчужины в любом жанре

```python
%timeit
print("ТЕСТ-ДРАЙВ: Ищем лучшие фильмы в жанре 'Drama' (Драма)")

top_drama = ratings_movies.top_by_genre('Drama', 10)
if top_drama and len(top_drama) > 0:
    print("\nТОП-10 ЛУЧШИХ ДРАМ ПО ВЕРСИИ ЗРИТЕЛЕЙ:\n")
    for i, (title, rating) in enumerate(top_drama.items(), 1):
        stars = '★' * int(float(rating))
        print(f"   {i:2d}. {title[:55]:55} → {rating} {stars}")
else:
    print("Фильмы в жанре 'Drama' не найдены или данные отсутствуют")
```

Звёздный рейтинг — Что любят смотреть пользователи?

Метод: dist_by_rating() - возвращает словарь {оценка: количество}, отсортированный по оценке.

```python
%timeit
print("РАСПРЕДЕЛЕНИЕ ОЦЕНОК ПО ЗНАЧЕНИЯМ")

rating_dist = ratings_movies.dist_by_rating()
if rating_dist:
    for rating, count in rating_dist.items():
        stars = '★' * int(float(rating))
        bar = '█' * (count // 20)
        print(f"{rating} {stars}: {count:4d} {bar}")
else:
    print("Нет данных")
```

Анализ распределения оценок:

-Позитивный уклон - пользователи чаще ставят высокие оценки (4-5 звёзд)
-Редкие оценки - 0.5 и 1.5 звёзды встречаются реже
-Оценки могут быть с шагом 0.5 (0.5 - 5.0)

ХИТ-ПАРАД — Какие фильмы получили больше всего оценок?

Метод: top_by_num_of_ratings(n) - возвращает топ-n фильмов с наибольшим количеством оценок.

```python
%timeit
print("ТОП-10 ФИЛЬМОВ ПО КОЛИЧЕСТВУ ОЦЕНОК")

top_by_count = ratings_movies.top_by_num_of_ratings(10)

if top_by_count and isinstance(top_by_count, dict) and len(top_by_count) > 0:
    for i, (title, count) in enumerate(top_by_count.items(), 1):
        if title is not None:
            bar = '█' * (count // 10) if count >= 10 else '█'
            print(f"{i:2d}. {str(title)[:50]:50} → {count:3d} оценок {bar}")
        else:
            print(f"{i:2d}. [Неизвестный фильм] → {count:3d} оценок")
else:
    print("Нет данных")
```

Анализ популярности:

Самые популярные фильмы (по количеству оценок):
-Известные хиты - фильмы, которые смотрело много людей
-Культовая классика - фильмы, которые пересматривают
-Современные блокбастеры - привлекают больше внимания

ЛУЧШИЕ ИЗ ЛУЧШИХ — Фильмы с высочайшим средним баллом
Посмотрим на фильмы, которые зрители просто обожают! Те самые шедевры, которые получают почти 5 звёзд.

Метод: top_by_ratings(n, metric='average') - возвращает топ-n фильмов по средней оценке.

```python
%timeit
print("ТОП-10 ФИЛЬМОВ ПО СРЕДНЕЙ ОЦЕНКЕ")

top_by_avg = ratings_movies.top_by_ratings(10, 'average')
if top_by_avg:
    for i, (title, rating) in enumerate(top_by_avg.items(), 1):
        stars = '★' * int(float(rating))
        print(f"{i:2d}. {title[:50]:50} → {rating} {stars}")
else:
    print("Нет данных")
```

БИТВА МНЕНИЙ — Самые спорные фильмы

А есть ли фильмы, которые вызывают споры? Те, которые одни обожают, а другие ненавидят? Давайте найдём их!

Метод: top_controversial(n) - возвращает топ-n фильмов с наибольшей дисперсией оценок.

```python
%timeit
print("ТОП-10 САМЫХ СПОРНЫХ ФИЛЬМОВ")

controversial = ratings_movies.top_controversial(10)
if controversial:
    for i, (title, variance) in enumerate(controversial.items(), 1):
        print(f"{i:2d}. {title[:50]:50} → дисперсия = {variance}")
else:
    print("Нет данных")
```

Анализ спорных фильмов:

Дисперсия показывает разброс мнений:
-Высокая дисперсия = мнения разделились (одни любят, другие ненавидят)
-Низкая дисперсия = все согласны с оценкой

Спорные фильмы часто:
-Экспериментальные или нестандартные
-Культовые с сильной любовью и ненавистью
-С неожиданными поворотами сюжета

АКТИВНЫЕ ЗРИТЕЛИ — Кто оставляет больше всего оценок?
Кто самый активный? Кто посмотрел больше всего фильмов?
Метод: dist_by_num_of_ratings() - распределение пользователей по активности.

```python
%timeit
print("РАСПРЕДЕЛЕНИЕ ПОЛЬЗОВАТЕЛЕЙ ПО АКТИВНОСТИ")

user_dist = ratings_users.dist_by_num_of_ratings()
if user_dist:
    sorted_users = sorted(user_dist.items(), key=lambda x: x[1], reverse=True)
    print(f"Всего пользователей: {len(sorted_users)}")
    print(f"\nТОП-10 САМЫХ АКТИВНЫХ ПОЛЬЗОВАТЕЛЕЙ:")
    for i, (user_id, count) in enumerate(sorted_users[:10], 1):
        bar = '█' * (count // 20)
        print(f"{i:2d}. Пользователь {user_id}: {count:3d} оценок {bar}")
    
    counts = [c for _, c in sorted_users]
    print(f"\nСредняя активность: {sum(counts) / len(counts):.1f} оценок")
    print(f"Максимум: {max(counts)} оценок")
    print(f"Минимум: {min(counts)} оценок")
else:
    print("Нет данных")
```

Распределение пользователей по среднему рейтингу

Метод: dist_by_rating(metric) - возвращает словарь {user_id: средний/медианный рейтинг}.

```python
%timeit
print("ТОП-10 ПОЛЬЗОВАТЕЛЕЙ ПО СРЕДНЕЙ ОЦЕНКЕ")

user_avg = ratings_users.dist_by_rating('average')
if user_avg:
    sorted_users = sorted(user_avg.items(), key=lambda x: x[1], reverse=True)[:10]
    for i, (user_id, avg) in enumerate(sorted_users, 1):
        stars = '★' * int(float(avg))
        print(f"{i:2d}. Пользователь {user_id}: {avg} {stars}")
else:
    print("Нет данных")


print("ТОП-10 ПОЛЬЗОВАТЕЛЕЙ ПО МЕДИАНЕ ОЦЕНОК")

user_median = ratings_users.dist_by_rating('median')
if user_median:
    sorted_users = sorted(user_median.items(), key=lambda x: x[1], reverse=True)[:10]
    for i, (user_id, med) in enumerate(sorted_users, 1):
        stars = '★' * int(float(med))
        print(f"{i:2d}. Пользователь {user_id}: {med} {stars}")
else:
    print("Нет данных")
```

Анализ пользовательских предпочтений:

-Строгие критики - ставят низкие оценки (2-3 звезды)
-Энтузиасты - почти всегда ставят 5 звёзд
-Объективные - используют весь диапазон оценок

ЭКСПЕРТЫ ИЛИ ЭНТУЗИАСТЫ? — У кого самый разный вкус?

Некоторые зрители любят почти всё, а другие — настоящие критики. Найдём тех, у кого самый широкий диапазон оценок!

Метод: top_users_with_biggest_variance(n) - пользователи с самым разбросанным мнением.

```python
%timeit
print("ТОП-10 ПОЛЬЗОВАТЕЛЕЙ С БОЛЬШОЙ ВАРИАТИВНОСТЬЮ ОЦЕНОК")

variable_users = ratings_users.top_users_with_biggest_variance(10)
if variable_users:
    for i, (user_id, variance) in enumerate(variable_users.items(), 1):
        print(f"{i:2d}. Пользователь {user_id}: дисперсия = {variance}")
else:
    print("Нет данных")
```

Анализ вариативности пользователей:

Пользователи с высокой вариативностью:
-Смотрят разные жанры - от шедевров до провалов
-Объективные критики - оценивают каждый фильм по достоинству
-Не боятся ставить низкие оценки - даже любимым жанрам

МНОГОСЛОВНЫЕ ТЕГИ — У кого богатый словарный запас?

Символы — это одно, а слова — другое! Кто использует больше всего слов в одном теге?

Метод: most_words(n) - возвращает топ-n тегов с максимальным количеством слов.

```python
%timeit
print("ТОП-10 ТЕГОВ С БОЛЬШИМ КОЛИЧЕСТВОМ СЛОВ")

wordy = tags.most_words(10)
if wordy:
    for i, (tag, word_count) in enumerate(wordy.items(), 1):
        preview = tag[:50] + '...' if len(tag) > 50 else tag
        print(f"{i:2d}. '{preview}' → {word_count} слов")
else:
    print("Нет данных")
```

Анализ:
Многословные теги показывают, что некоторые пользователи используют теги как мини-рецензии.

САМЫЕ ДЛИННЫЕ ТЕГИ — Кто пишет целые сочинения?

Есть пользователи, которые не жалеют символов! Давайте найдём самые длинные теги-описания.

Метод: longest(n) - возвращает топ-n тегов с максимальным количеством символов.

```python
%timeit
print("ТОП-10 САМЫХ ДЛИННЫХ ТЕГОВ")

longest = tags.longest(10)
if longest:
    for i, tag in enumerate(longest, 1):
        print(f"{i:2d}. {tag[:65]}... ({len(tag)} симв.)")
else:
    print("Нет данных")
```

Анализ:
Длинные теги часто содержат:
-Целые предложения и описания
-Личные впечатления от просмотра
-Контекст ("watch with family", "better than the book")

ИДЕАЛЬНЫЕ ТЕГИ — Длинные И многословные

А теперь найдём те теги, которые одновременно длинные и содержат много слов — настоящие шедевры тегирования!

Метод: most_words_and_longest(n) - возвращает теги, которые входят в оба топа.

```python
%timeit
print("ТЕГИ, КОТОРЫЕ ОДНОВРЕМЕННО ДЛИННЫЕ И МНОГОСЛОВНЫЕ")

intersection = tags.most_words_and_longest(15)
if intersection:
    for i, tag in enumerate(intersection, 1):
        print(f"{i:2d}. {tag[:70]}")
else:
    print("Нет пересекающихся тегов")
```

Анализ:

Теги, которые одновременно длинные и многословные - это самые подробные пользовательские описания.

ХИТ-ПАРАД ТЕГОВ — Что пишут чаще всего?

Давайте посмотрим на самые популярные слова и фразы, которые оставляют пользователи!

Метод: most_popular(n) - возвращает топ-n тегов по частоте использования.


```python
%timeit
print("ТОП-10 САМЫХ ПОПУЛЯРНЫХ ТЕГОВ")

popular = tags.most_popular(10)
if popular:
    for i, (tag, count) in enumerate(popular.items(), 1):
        bar = '█' * (count // 2)
        print(f"{i:2d}. {tag:25} → {count:3d} раз {bar}")
else:
    print("Нет данных")
```

Анализ:

Самые популярные теги показывают, что пользователи чаще всего отмечают:
-Эмоции: funny, sad
-Атмосферу: dark, suspenseful, intense  
-Качество: watch again, worth watching
-Жанры: documentary, thriller

ПОИСК ПО СМЫСЛУ — Теги с определёнными словами

*Давайте поищем теги, связанные с конкретными темами. Что пишут про "great" и "bad"?

Метод: tags_with(word) - возвращает все уникальные теги, содержащие заданное слово.

```python
%timeit
print("ТЕГИ, СОДЕРЖАЩИЕ СЛОВО 'good'")

good_tags = tags.tags_with('good')
if good_tags:
    for i, tag in enumerate(good_tags[:10], 1):
        print(f"{i:2d}. {tag}")
    if len(good_tags) > 10:
        print(f"... и ещё {len(good_tags) - 10} тегов")
else:
    print("Теги не найдены")

print("ТЕГИ, СОДЕРЖАЩИЕ СЛОВО 'bad'")

bad_tags = tags.tags_with('bad')
if bad_tags:
    for i, tag in enumerate(bad_tags[:10], 1):
        print(f"{i:2d}. {tag}")
else:
    print("Теги не найдены")
```

Анализ поиска:
-good→ чаще всего позитивные оценки
-bad→ негативные отзывы

Получение данных с IMDb (OMDb API)

Метод: get_imdb(list_of_movies, list_of_fields) - получает данные для списка фильмов.

```python
%timeit
print("ПОЛУЧЕНИЕ ДАННЫХ С OMDb API")

sample_movies = [links.data[i][1] for i in range(min(3, len(links.data)))]
fields = ['Title', 'Year', 'Director', 'Runtime', 'imdbRating']

print(f"📋 Запрашиваем фильмы: {sample_movies}")
print(f"🔍 Поля: {fields}")
print()

imdb_data = links.get_imdb(sample_movies, fields)
if imdb_data:
    for movie_data in imdb_data:
        print(f"🎬 Фильм {movie_data[0]}:")
        for i, field in enumerate(fields, 1):
            print(f"   {field}: {movie_data[i]}")
        print()
else:
    print("Не удалось получить данные с OMDb API")
```

Анализ данных из OMDb:

Метод get_imdb() позволяет получать:
-Базовую информацию о фильме (название, год)
-Режиссёра - кто создал фильм
-Хронометраж - длительность в минутах
-Рейтинг IMDb - оценка зрителей

МАСТЕРА РЕЖИССУРЫ — Кто снял больше всего фильмов?

Кто самый продуктивный режиссёр? Чьи фильмы чаще всего попадают в датасет?

Метод: top_directors(n) - возвращает топ-n режиссёров с наибольшим количеством фильмов.

```python
%timeit
print("ТОП-10 РЕЖИССЁРОВ ПО КОЛИЧЕСТВУ ФИЛЬМОВ")

top_dirs = links.top_directors(10)
if top_dirs:
    for i, (director, count) in enumerate(top_dirs.items(), 1):
        bar = '█' * count
        print(f"{i:2d}. {director:30} → {count} фильмов {bar}")
else:
    print("Нет данных или ошибка API")
```

Анализ режиссёров:

Топ режиссёров показывает:
-Самых продуктивных создателей фильмов
-Мастеров жанров (например, режиссёры комедий или драм)
-Современных и классических** режиссёров

Данные берутся из OMDb API и кэшируются для ускорения работы.

САМЫЕ ДОРОГИЕ ФИЛЬМЫ — Где бюджет зашкаливает?

Какой фильм стоил больше всего? Голливудские блокбастеры или неожиданные претенденты?

Метод: most_expensive(n) - возвращает топ-n фильмов с наибольшим бюджетом.

```python
%timeit
print("ТОП-10 САМЫХ ДОРОГИХ ФИЛЬМОВ")

expensive = links.most_expensive(10)
if expensive:
    for i, (title, budget) in enumerate(expensive.items(), 1):
        # Форматируем бюджет для читаемости
        budget_formatted = f"${budget:,}" if budget else "N/A"
        print(f"{i:2d}. {title[:45]:45} → {budget_formatted}")
else:
    print("Нет данных о бюджетах")
```

Анализ бюджетов:

Самые дорогие фильмы обычно:
-Блокбастеры с высокими производственными затратами
-Фантастика и экшн - требуют много спецэффектов
-Фильмы с известными актёрами - высокие гонорары


САМЫЕ ПРИБЫЛЬНЫЕ — Кто окупился лучше всех?

Большой бюджет не всегда гарантирует успех. А какой фильм оказался самым прибыльным? Считаем сборы минус затраты!

Метод: most_profitable(n) - возвращает топ-n фильмов по разнице между сборами и бюджетом.

```python
%timeit
print("ТОП-10 САМЫХ ПРИБЫЛЬНЫХ ФИЛЬМОВ")

profitable = links.most_profitable(10)
if profitable:
    for i, (title, profit) in enumerate(profitable.items(), 1):
        profit_formatted = f"${profit:,}" if profit else "N/A"
        print(f"{i:2d}. {title[:45]:45} → {profit_formatted}")
else:
    print("Нет данных о прибыли")
```

Анализ прибыльности:

Прибыль = Сборы - Бюджет

Самые прибыльные фильмы:
-Не всегда самые дорогие - важна окупаемость
-Часто неожиданные хиты - малобюджетные с большими сборами
-Франшизы - имеют встроенную аудиторию

КИНО-МАРАФОНЦЫ — Самые длинные фильмы

Кто готов к марафонскому просмотру? Найдём фильмы, которые длятся 3+ часа!

Метод: longest(n) - возвращает топ-n фильмов по длительности.

```python
%timeit
print("ТОП-10 САМЫХ ДЛИННЫХ ФИЛЬМОВ")

longest_movies = links.longest(10)
if longest_movies:
    for i, (title, runtime) in enumerate(longest_movies.items(), 1):
        hours = runtime // 60
        minutes = runtime % 60
        print(f"{i:2d}. {title[:45]:45} → {runtime} мин ({hours}ч {minutes}м)")
else:
    print("Нет данных о длительности")
```

Анализ длительности:

Самые длинные фильмы:
-Эпические драмы и исторические - требуют времени для повествования
-Режиссёрские версии - содержат дополнительный материал
-Классические фильмы - имели другой темп повествования

ЦЕНА МИНУТЫ — Какой фильм самый дорогой в пересчёте на минуту?

Сколько стоит одна минута экранного времени? Это метрика, которая уравнивает короткие и длинные фильмы!

Метод: top_cost_per_minute(n) - возвращает топ-n фильмов по стоимости минуты (бюджет / длительность).

```python
%timeit
print("ТОП-10 ФИЛЬМОВ ПО СТОИМОСТИ МИНУТЫ")

cost_per_min = links.top_cost_per_minute(10)
if cost_per_min:
    for i, (title, cost) in enumerate(cost_per_min.items(), 1):
        print(f"{i:2d}. {title[:45]:45} → ${cost:,.2f} в минуту")
else:
    print("Нет данных для расчёта")
```

Анализ стоимости минуты:

Этот метрика показывает:
-Насколько дорого обходится каждая минута экранного времени
-Самые затратные жанры (обычно экшн и фантастика)
-Эффективность производства - низкая стоимость не всегда плохо
