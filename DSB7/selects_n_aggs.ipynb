```python
import pandas as pd
```

## Загружаем JSON файл, созданный в предыдущем упражнении

```python
df = pd.read_json('../data/auto.json', orient='records')
```

## Установка CarNumber в качестве индекса

```python
df.set_index('CarNumber', inplace=True)
```

## Выборки (фильтрация)

```python
# 1. Отображаем строки, где штрафы больше 2100
fines_gt_2100 = df[df['Fines'] > 2100]
print(fines_gt_2100)
```

```python
# 2. Отображаем строки, где штрафы больше 2100, а возврат равен 2
fines_gt_2100_refund_2 = df[(df['Fines'] > 2100) & (df['Refund'] == 2)]
print(fines_gt_2100_refund_2)
```

```python
# 3. Отображаем строки, где модель входит в список ['Focus', 'Corolla']
focus_corolla = df[df['Model'].isin(['Focus', 'Corolla'])]
print(focus_corolla)
```

```python
# 4. Отображаем строки, где номер автомобиля входит в список
selected_cars = df[df.index.isin(['Y7689C197RUS', '92928M178RUS', '7788KT197RUS', 'H115YO163RUS', 'X758HY197RUS'])]
print(selected_cars)
```

## Агрегации по Make и Model

```python
# 1. Медианные значения штрафов, сгруппированные по Make
median_fines_by_make = df.groupby('Make')['Fines'].median()
print(median_fines_by_make)
```

```python
# 2. Медианные значения штрафов, сгруппированные по Make и Model
median_fines_by_make_model = df.groupby(['Make', 'Model'])['Fines'].median()
print(median_fines_by_make_model)
```

```python
# 3. Количество штрафов, сгруппированное по Make и Model
count_fines_by_make_model = df.groupby(['Make', 'Model'])['Fines'].count()
print(count_fines_by_make_model)
print("\nДля доверия к медиане нужно достаточно наблюдений в каждой группе")
print("Группы с малым количеством записей (< 10) дают менее надежную медиану")
print(f"Группы с менее чем 10 записями: {(count_fines_by_make_model < 10).sum()} из {len(count_fines_by_make_model)}\n")
```

```python
# 4. Минимальные и максимальные штрафы по Make и Model
min_max_fines_by_make_model = df.groupby(['Make', 'Model'])['Fines'].agg(['min', 'max'])
print(min_max_fines_by_make_model)
```

```python
# 5. Стандартное отклонение штрафов по Make и Model
std_fines_by_make_model = df.groupby(['Make', 'Model'])['Fines'].std()
print(std_fines_by_make_model)
```

## Агрегации по номерам автомобилей

```python
# 1. Номера автомобилей, сгруппированные по количеству штрафов в порядке убывания
fines_count_by_car = df.groupby(df.index)['Fines'].count().sort_values(ascending=False)
print(fines_count_by_car)
```

```python
# 2. Все строки, соответствующие топ-1 номеру по количеству штрафов
top1_by_count = df[df.index == fines_count_by_car.index[0]]
print(top1_by_count)
```

```python
# 3. Номера автомобилей, сгруппированные по сумме штрафов в порядке убывания
fines_sum_by_car = df.groupby(df.index)['Fines'].sum().sort_values(ascending=False)
print(fines_sum_by_car)
```

```python
# 4. Все строки, соответствующие топ-1 номеру по сумме штрафов
top1_by_sum = df[df.index == fines_sum_by_car.index[0]]
print(top1_by_sum)
```

## Проверка: номера автомобилей, привязанные к разным моделям

```python
# 5. Таблица: номера автомобилей, привязанные к разным моделям
# Группируем по номеру автомобиля и собираем уникальные модели
cars_with_multiple_models = df.groupby(df.index)['Model'].nunique()
cars_multiple_models = cars_with_multiple_models[cars_with_multiple_models > 1]

print("Номера автомобилей, привязанные к разным моделям:")
if len(cars_multiple_models) > 0:
    print(cars_multiple_models)
    
    print("\nДетали по каждому номеру:")
    for car_number in cars_multiple_models.index:
        print(f"\nНомер: {car_number}")
        car_data = df[df.index == car_number]
        print(f"  Модели: {car_data['Model'].unique()}")
        print(f"  Марки: {car_data['Make'].unique()}")
        print(f"  Количество записей: {len(car_data)}")
else:
    print("Нет номеров автомобилей, привязанных к разным моделям")
    print("Каждый номер автомобиля связан только с одной моделью")
```
