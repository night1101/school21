```python
import pandas as pd
```

## Загрузка данных из log-файла

```python
df = pd.read_csv(
    '../data/feed-views.log',
    sep='\t',
    skiprows=[2,3],
    skipfooter=2,
    names=['datetime', 'user'],
    index_col='datetime',
    engine='python'
)
```

```python
print("Первые 5 строк:")
print(df.head())
print("Последние 5 строк:")
print(df.tail())
```

## Переименование индекса и сохранение

```python
df.index.name='date_time'
```

## Сохраняем в файл с разделителем ';'

```python
df.to_csv('../data/feed-views-semicolon.log', sep=';')
```
